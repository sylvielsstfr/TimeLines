```python
import plotly.graph_objects as go

# Data (approximate fractions from PSTN-056)
labels = [
    "Wide-Fast-Deep (WFD)",
    "Deep Drilling Fields (DDF)",
    "Target of Opportunity (ToO)",
    "Mini/Micro-surveys"
]

values = [85, 6.5, 3.5, 5]  # percentages (approximate but consistent)

# Create pie chart
fig = go.Figure(data=[go.Pie(
    labels=labels,
    values=values,
    hole=0.4,  # donut style (cleaner for presentations)
    textinfo='label+percent',
    insidetextorientation='radial'
)])

fig.update_layout(
    title="Rubin Observatory LSST Time Allocation",
    font=dict(size=16),
    width=800,
    height=800
)


fig.show()
```

```python
import matplotlib.pyplot as plt

labels = [
    "WFD",
    "DDF",
    "ToO",
    "Mini/Micro"
]

sizes = [85, 6.5, 3.5, 5]

fig, ax = plt.subplots()

ax.pie(
    sizes,
    labels=labels,
    autopct='%1.1f%%',
    startangle=90
)

ax.set_title("Rubin LSST Observing Time Allocation")

plt.show()
```
