```python
import pandas as pd

# Structure de la timeline avec des étapes fictives que tu peux modifier
timeline_data = [
    # date_debut, date_fin, étape, colonne (axe)
    ("2025-01","2027-12", "Calibration instrumentale : TCBP (ZTF, Auxtel Rubin obs)","Instrumentation"),
    ("2025-01","2027-12", "Paramètres atmosphériques, FGCM/Auxtel et corr. photom. courbes lum.","Auxtel"),
    ("2026-02","2029-12", "Analyse des alertes, détection de SN","Broker Fink"),
    ("2025-04", "2029-12", "Analyse des données Rubin et science des Supernovae", "Rubin Observatory"),
    ("2025-01", "2029-12", "Activités scientifiques relative à DESC", "DESC"),
    ("2025-06", "2025-07", "DP1 (Data LSSTComCam)", "Data Previews"),
    ("2026-09", "2026-10", "Early DP2 (Data LSSTCam)","Data Previews"),
    ("2026-12", "2027-01", "Full DP2 (Data LSSTCam)","Data Previews"),    
    ("2027-09", "2027-12", "DR1 (Data LSSTCam, 6 months) ", "Data Releases"),    
]

# Création du DataFrame
df_timeline = pd.DataFrame(timeline_data, columns=["start", "end", "task", "category"])

# Conversion des dates en datetime
df_timeline["start"] = pd.to_datetime(df_timeline["start"])
df_timeline["end"] = pd.to_datetime(df_timeline["end"])

print(df_timeline)
```

```python
df_timeline
```

```python
import plotly.express as px

fig = px.timeline(df_timeline, x_start="start", x_end="end", y="category", color="task", title="Timeline 2025–2030")
fig.update_yaxes(autorange="reversed")  # Pour avoir Auxtel en haut
fig.show()
```

```python
fig = px.timeline(
    df_timeline,
    x_start="start",
    x_end="end",
    y="category",
    color="category",
    hover_name="task",
    title="Nos activités"
)


fig.update_yaxes(autorange="reversed")

fig.update_layout(
    title="Rubin-LSST Timeline 2025-2030",
    height=800,
    width=1600,
    margin=dict(l=80, r=50, t=80, b=120),

    legend=dict(
        orientation="h",
        yanchor="top",
        y=-0.2,
        xanchor="center",
        x=0.5
    ),

    title_font=dict(size=24),
    xaxis_title_font=dict(size=20),
    yaxis_title_font=dict(size=20),
    legend_title_font=dict(size=18),
    legend_font=dict(size=14),
    font=dict(size=16)
)

fig.update_xaxes(dtick="M6", tickangle=45)

fig.show()
```

```python
fig = px.timeline(
    df_timeline,
    x_start="start",
    x_end="end",
    y="category",
    color="category",
    text="task",
    title="Nos activités"
)


#fig.update_yaxes(autorange="reversed")




fig.update_layout(
    title="Nos activités de 2025-2030",
    height=800,
    width=1600,
    margin=dict(l=80, r=50, t=80, b=120),

    legend=dict(
        orientation="h",
        yanchor="top",
        y=-0.2,
        xanchor="center",
        x=0.5
    ),

    title_font=dict(size=24),
    xaxis_title_font=dict(size=20),
    yaxis_title_font=dict(size=20),
    legend_title_font=dict(size=18),
    legend_font=dict(size=14),
    font=dict(size=16)
)

fig.update_xaxes(dtick="M6", tickangle=45)

fig.update_traces(
    textposition="inside",
    insidetextanchor="middle",
    textfont=dict(size=18)
)


fig.show()
```
