```python
import plotly.express as px

fig = px.timeline(df_timeline, x_start="start", x_end="end", y="member", color="status", title="Timeline Rubin-LSST project :2024–2035")
fig.update_yaxes(autorange="reversed")  # Pour avoir Auxtel en haut
fig.show()
```
