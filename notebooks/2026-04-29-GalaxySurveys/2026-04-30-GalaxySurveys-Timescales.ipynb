```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib.lines import Line2D

# Configuration de la figure
fig, ax = plt.subplots(figsize=(14, 5))
ax.set_xlim(2010, 2035)
ax.set_ylim(-0.5, 4.5)  # Ajustement pour éviter le rognage

# Suppression des axes inutiles (mais on garde les années)
ax.set_xticks(range(2010, 2036, 2))
ax.set_xticklabels(range(2010, 2036, 2), fontsize=10, rotation=45, ha='right')
ax.set_yticks([])  # Pas d'axe y visible
ax.set_title("Galaxy Surveys Timeline: Stages II–IV", fontsize=16, pad=20)

# Ajout des lignes verticales pour les années (grises et discrètes)
for year in range(2010, 2036, 2):
    ax.axvline(x=year, color='gray', linestyle='--', linewidth=0.5, alpha=0.5)

# Définition des relevés avec leurs dates, couleurs et positions
surveys = [
    # Stage II (bleu)
    {"name": "DES", "start": 2013, "end": 2019, "y": 3.5, "color": "#1f77b4"},
    {"name": "eBOSS", "start": 2014, "end": 2020, "y": 2.8, "color": "#1f77b4"},

    # Stage III (orange)
    {"name": "DESI", "start": 2021, "end": 2026, "y": 1.8, "color": "#ff7f0e"},
    {"name": "J-PAS", "start": 2020, "end": 2025, "y": 1.1, "color": "#ff7f0e"},

    # Stage IV (violet)
    {"name": "Euclid", "start": 2024, "end": 2030, "y": 0.4, "color": "#9467bd"},
    {"name": "LSST", "start": 2026, "end": 2036, "y": -0.3, "color": "#9467bd"},
]

# Ajout des barres pour chaque relevé
for survey in surveys:
    ax.add_patch(
        patches.Rectangle(
            (survey["start"], survey["y"] - 0.25),
            survey["end"] - survey["start"],
            0.5,
            linewidth=1,
            edgecolor='black',
            facecolor=survey["color"],
            alpha=0.8
        )
    )
    # Ajout du nom du relevé au centre de la barre (en blanc)
    ax.text(
        (survey["start"] + survey["end"]) / 2,
        survey["y"],
        survey["name"],
        ha='center',
        va='center',
        color='white',
        fontsize=12,
        fontweight='bold'
    )

# Légende pour les Stages
legend_elements = [
    Line2D([0], [0], marker='s', color='w', label='Stage II',
           markerfacecolor='#1f77b4', markersize=10),
    Line2D([0], [0], marker='s', color='w', label='Stage III',
           markerfacecolor='#ff7f0e', markersize=10),
    Line2D([0], [0], marker='s', color='w', label='Stage IV',
           markerfacecolor='#9467bd', markersize=10),
]
ax.legend(handles=legend_elements, loc='upper right', fontsize=12, framealpha=1)

# Ajustement des marges pour éviter le rognage
plt.tight_layout()
plt.subplots_adjust(bottom=0.2)  # Espace pour les années en bas

# Sauvegarde en PNG haute résolution
plt.savefig("galaxy_surveys_timeline_corrected.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib.lines import Line2D

# Configuration de la figure (carrée)
fig, ax = plt.subplots(figsize=(4, 4))
ax.set_xlim(-1.5, -0.5)
ax.set_ylim(-0.5, 0.5)
ax.set_aspect('equal')  # Pour un graphique carré
ax.set_xlabel(r"$w_0$", fontsize=14)
ax.set_ylabel(r"$w_a$", fontsize=14)
ax.set_title(r"Dark Energy Constraints: $w_0$ vs $w_a$ (95\% CL)", fontsize=10, pad=20)
ax.grid(True, linestyle='--', alpha=0.5)

# Modèle fiduciel (w0=-1, wa=0)
fiducial_w0, fiducial_wa = -1.0, 0.0

# Paramètres des ellipses pour chaque Stage (sigma_w0, sigma_wa, couleur)
# Basé sur les valeurs du DETF (Tableau 4 du rapport)
ellipse_params = {
    "Stage II": {"sigma_w0": 0.15, "sigma_wa": 0.4, "color": "#1f77b4"},
    "Stage III": {"sigma_w0": 0.04, "sigma_wa": 0.15, "color": "#ff7f0e"},
    "Stage IV": {"sigma_w0": 0.015, "sigma_wa": 0.07, "color": "#9467bd"},
}

# Dessiner les ellipses (95% CL ~ 2.45 sigma pour une distribution 2D gaussienne)
for stage, params in ellipse_params.items():
    sigma_w0 = params["sigma_w0"]
    sigma_wa = params["sigma_wa"]
    color = params["color"]

    # Calcul des axes de l'ellipse
    a = 2.45 * sigma_w0  # Demi-grand axe (w0)
    b = 2.45 * sigma_wa  # Demi-petit axe (wa)

    # Créer l'ellipse (centrée sur le modèle fiduciel)
    ellipse = patches.Ellipse(
        (fiducial_w0, fiducial_wa),
        width=2 * a,
        height=2 * b,
        linewidth=2,
        edgecolor=color,
        facecolor='none',
        label=stage
    )
    ax.add_patch(ellipse)

    # Ajouter le texte pour le Stage
    ax.text(
        fiducial_w0 + 0.05,
        fiducial_wa + 0.05 + (0.1 * list(ellipse_params.keys()).index(stage)),
        stage,
        color=color,
        fontsize=12,
        fontweight='bold'
    )

# Ajouter des lignes pointillées pour le modèle fiduciel (w0=-1, wa=0)
ax.axvline(x=fiducial_w0, color='black', linestyle=':', linewidth=1, alpha=0.7)
ax.axhline(y=fiducial_wa, color='black', linestyle=':', linewidth=1, alpha=0.7)

# Légende
ax.legend(loc='upper left', fontsize=8, framealpha=1)

# Sauvegarde en PNG haute résolution
plt.tight_layout()
plt.savefig("dark_energy_ellipses.png", dpi=300, bbox_inches='tight')
plt.show()
```
