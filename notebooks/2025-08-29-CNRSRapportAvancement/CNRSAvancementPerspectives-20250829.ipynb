```python
import matplotlib.pyplot as plt
import pandas as pd
import matplotlib.dates as mdates
from matplotlib.patches import Patch
import matplotlib.cm as cm

# Tes données
df_timeline = pd.DataFrame([
    ("2021-01","2021-12", "Cosmo & IA","DESC"),
    ("2022-01","2025-10", "Cosmo & IA + PZ (thesis)","DESC"),
    ("2025-10","2029-01", "DATA-Observations analysis ","DESC"),
    ("2021-01", "2022-06", "Comissioning", "Auxtel"),
    ("2022-09", "2025-02", "Pre-Observations", "Auxtel"),
    ("2025-05", "2037-01", "Run-Observations", "Auxtel"),
#    ("2021-01", "2024-12", "Delegate Assembly", "Prep Data"),
#    ("2025-01", "2025-06", "Rubin Science Assembly", "Prep Data"),
    ("2024-04", "2024-05","LSSTComCamSim rehearsal","Commissioning"),
    ("2024-11", "2024-12","LSSTComCam on sky","Commissioning"),
    ("2025-04", "2025-07","LSSTCam on sky","Commissioning"),
    ("2025-07", "2025-12", "LSSTCam Eng. Opt & Science validation", "Full Survey"),
    ("2026-01", "2027-01", "LSSTCam Y1 Survey", "Full Survey"),
    ("2027-01", "2037-01", "LSSTCam Survey", "Full Survey"),
    ("2021-01", "2025-07","Fink preparation","Fink Broker"),
    ("2025-07", "2025-09","Fink First Alerts Tests","Fink Broker"),
    ("2025-09", "2037-01","Fink Regular Alerts","Fink Broker"),
    ("2023-06", "2023-07", "DP0 (Simu Como-DC2)", "Data Preview"),
    ("2025-06", "2025-07", "DP1 (Data LSSTComCam)", "Data Preview"),
    ("2026-03", "2026-05", "DP2 (Data LSSTCam, 6 months)", "Data Preview"),
    ("2026-09", "2027-01", "DR1 (Data LSSTCam, 6 months) ", "Data Release"),
    ("2027-09", "2028-01", "DR2 (Data LSSTCam)", "Data Release"),
    ("2028-09", "2029-01", "DR3 (Data LSSTCam)", "Data Release"),
], columns=["Start", "End", "Task", "Category"])



df_timeline["Start"] = pd.to_datetime(df_timeline["Start"])
df_timeline["End"] = pd.to_datetime(df_timeline["End"])

# Génération des couleurs par tâche
#cmap = cm.get_cmap('Set3', len(df_timeline))
cmap = plt.colormaps.get_cmap('Set3').resampled(len(df_timeline))

colors = [cmap(i) for i in range(len(df_timeline))]
task_colors = {task: color for task, color in zip(df_timeline['Task'], colors)}

# Position des catégories sur l’axe X
categories = df_timeline['Category'].unique()
category_pos = {cat: i for i, cat in enumerate(categories)}

# Figure
fig, ax = plt.subplots(figsize=(12, 12))

# Tracer les barres verticales
for idx, row in df_timeline.iterrows():
    x = category_pos[row['Category']]
    y_start = row['Start']
    y_end = row['End']
    mid_date = y_start + (y_end - y_start) / 2
    color = task_colors[row['Task']]

    # Barre verticale
    ax.plot([x, x], [y_start, y_end], linewidth=30, color=color)
    
    # Texte centré
    ax.text(x, mid_date, row['Task'], ha='center', va='center', fontsize=16, rotation=0,color='black')

# Axe X = catégories
ax.set_xticks(list(category_pos.values()))
ax.set_xticklabels(list(category_pos.keys()), rotation=45)

# Axe Y = dates
ax.yaxis_date()
ax.yaxis.set_major_formatter(mdates.DateFormatter('%Y-%m'))
ax.invert_yaxis()  # Pour avoir 2021 en haut

# Légende
legend_elements = [Patch(facecolor=color, label=task) for task, color in task_colors.items()]
ax.legend(handles=legend_elements, title="Tâches", bbox_to_anchor=(1.05, 1), loc='upper left')

# Finalisation
plt.title("Timeline verticale par activité (2021–2030)")
plt.ylabel("Date")
plt.tight_layout()
plt.show()
```
